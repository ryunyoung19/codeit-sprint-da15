```python
import re

def split_dates(value):
    if pd.isna(value):
        return value, None
    # YYYY-MM-DD 형태 날짜를 모두 찾기
    dates = re.findall(r"\d{4}-\d{2}-\d{2}", str(value))
    if len(dates) >= 2:
        return dates[0], dates[1]  # 최초개봉일, 재개봉일
    elif len(dates) == 1:
        return dates[0], None
    else:
        return value, None

df_movie_info[["개봉일_최초", "재개봉일"]] = df_movie_info["개봉"].apply(
    lambda x: pd.Series(split_dates(x))
)

df_movie_info[df_movie_info["재개봉일"].notna()][["title", "개봉", "개봉일_최초", "재개봉일"]]
```
